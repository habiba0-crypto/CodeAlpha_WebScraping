```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

print("Libraries imported successfully!")
```

```python
import requests

url = "https://books.toscrape.com/"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)
```

```python
books = soup.find_all("article", class_="product_pod")

print("Total Books Found:", len(books))

for book in books:
    title = book.h3.a["title"]
    print(title)
```

```python
book_data = []

for book in books:

    title = book.h3.a["title"]

    price = book.find("p", class_="price_color").text

    rating = book.find("p", class_="star-rating")["class"][1]

    availability = book.find(
        "p", class_="instock availability"
    ).text.strip()

    book_data.append({
        "Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

print("Data extracted successfully!")
```

```python
df = pd.DataFrame(book_data)

df.head(20)
```

```python
df.to_csv("CodeAlpha_WebScraping.csv", index=False)

print("CSV file saved successfully!")
```

```python
all_books = []

for page in range(1, 51):

    url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:

        title = book.h3.a["title"]

        price = book.find("p", class_="price_color").text

        rating = book.find("p", class_="star-rating")["class"][1]

        availability = book.find(
            "p", class_="instock availability"
        ).text.strip()

        all_books.append({
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": availability
        })

print("Total Books Scraped:", len(all_books))
```

```python
df_all = pd.DataFrame(all_books)

df_all.head(10)
```

```python
df_all.to_csv("CodeAlpha_WebScraping.csv", index=False)

print("Complete dataset saved successfully!")
```

```python
# Remove unwanted currency symbols and characters

df_all["Price"] = df_all["Price"].str.replace(
    "Â", "", regex=False
)

df_all["Price"] = df_all["Price"].str.replace(
    "£", "", regex=False
)

# Convert price into numeric format

df_all["Price"] = pd.to_numeric(
    df_all["Price"],
    errors="coerce"
)

# Display first 5 records

print(df_all[["Title", "Price"]].head())
```

```python
print(df_all["Price"].dtype)
print(df_all["Price"].isnull().sum())
```

```python
rating_mapping = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df_all["Rating"] = df_all["Rating"].map(rating_mapping)

print(df_all[["Title", "Rating"]].head(10))
```

```python
print("Dataset Shape:", df_all.shape)

print("\nData Types:")
print(df_all.dtypes)

print("\nMissing Values:")
print(df_all.isnull().sum())

print("\nDuplicate Records:")
print(df_all.duplicated().sum())
```

```python
df_all.to_csv(
    "CodeAlpha_WebScraping_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
